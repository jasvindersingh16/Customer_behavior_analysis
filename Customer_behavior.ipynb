```python
cd['Review Rating']=cd.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
cd.isnull().sum()
```

```python
cd.columns=cd.columns.str.lower()
cd.columns = cd.columns.str.replace(' ', '_')
cd.rename(columns={'purchase_amount_(usd)':'purchase_amount'},inplace=True)
```

```python
cd.columns
```

```python
bins = [0, 18, 35, 60, 100]
labels = ['Young-Adult', 'Adult', 'Middle-aged', 'Senior']

cd['age_group'] = pd.cut(
    cd['age'],
    bins=bins,
    labels=labels,
    right=True
)
```

```python
cd[['age','age_group']].head(10)
```

```python
# create new column purchase_frequency_days

frequency_mapping={'Weekly':7,
                   'Fortnightly':14,
                   'Monthly':30,
                   'Bi-Weekly':14,
                   'Quarterly':90,
                   'Annually':365,
                   'Every 3 Months':90
                   }
cd['purchase_frequency_days']=cd['frequency_of_purchases'].map(frequency_mapping)
```

```python
cd[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
cd[['discount_applied','promo_code_used']].head(10)
```

```python
cd=cd.drop('promo_code_used',axis=1)
```
